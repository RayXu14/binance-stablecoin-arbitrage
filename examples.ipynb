```python
import binance
from binance.spot import Spot
from binance_api_ks import key, secret

client = Spot(base_url='https://api1.binance.com',
              api_key=key, api_secret=secret)
# https://api2.binance.com
# https://api3.binance.com

# Get account and balance information
client.account()
```

```python
# Post a new order
params = {
    'symbol': 'FDUSDUSDT',
    'side': 'BUY',
    'type': 'LIMIT',
    'timeInForce': 'GTC',
    'quantity': 10,
    'price': 0.9999
}

try:
    response = client.new_order(**params)
    print(response)
except binance.error.ClientError as e:
    if e.status_code == 400 and e.error_code == -1013:
        print('[ERROR] Tried to create an order with a quantity lower than the minimun required')
    else:
        raise e

```
